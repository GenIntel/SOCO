# Explore SOCO

Load deterministic intra- and cross-category pairs and visualize exact semantic (SOC) and one-to-many concept (CC) matches. Set `SOCO_ROOT` before starting the notebook, or edit `DATA_ROOT` below.

```python
import os
from pathlib import Path
import matplotlib.pyplot as plt
from soco import SOCOPairs, plot_pair

DATA_ROOT = Path(os.environ.get("SOCO_ROOT", "data/SOCOv1"))
DATA_ROOT
```

```python
intra = SOCOPairs(DATA_ROOT, pair_type="intra", max_pairs_per_category=1)
cross = SOCOPairs(DATA_ROOT, pair_type="cross", max_pairs_per_category=1)
print(f"intra: {len(intra)} loaded index entries across {len(intra.categories)} categories")
print(f"cross: {len(cross)} loaded index entries across {len(cross.categories)} pair groups")
```

## Intra-category pair

```python
intra_sample = intra[1]
print(intra_sample.annotation_path)
print(f"semantic matches: {len(intra_sample.semantic_matches)}")
print(f"concept groups: {len(intra_sample.concept_matches)}")
plot_pair(intra_sample, mode="semantic")
plt.show()
plot_pair(intra_sample, mode="concept")
plt.show()
```

## Cross-category (inter-class) pair

```python
cross_sample = cross[1]
print(f"{cross_sample.source.category} -> {cross_sample.target.category}")
print(cross_sample.annotation_path)
plot_pair(cross_sample, mode="semantic")
plt.show()
plot_pair(cross_sample, mode="concept")
plt.show()
```
